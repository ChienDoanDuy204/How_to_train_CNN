```python
import torch
import torch.nn as nn
```

```python
block1 = nn.Sequential(
    nn.Conv2d(in_channels=3, out_channels=64, padding='same', stride=1, kernel_size=3),
    nn.Conv2d(in_channels=64, out_channels=64, padding='same', stride=1, kernel_size=3),
    nn.Conv2d(in_channels=64, out_channels=64, padding='same', stride=1, kernel_size=3),
    nn.MaxPool2d(kernel_size=2, stride=2)
)
# shortcut để thực hiện match dimentional khi thực hiện skip connection
shortcut = nn.Sequential(
    nn.Conv2d(in_channels=3, out_channels=64, padding=1, stride = 2, kernel_size=3),
    nn.BatchNorm2d(num_features=64)
)
```

```python
x = torch.rand((64,3,28,28))
```

```python
identity = shortcut(x)
```

```python
y = identity + block1(x)
```
