```python
import torch
import torch.nn as nn
from torchsummary import summary
```

```python
model = nn.Sequential(
    nn.Conv2d(in_channels=1,kernel_size=3,out_channels=64,padding='same', stride = 1),
    nn.ReLU(),
    nn.BatchNorm2d(num_features=64), # Sử dụng sau lớp Conv2d + activation/ không có
    nn.MaxPool2d(kernel_size=2, stride=2),
    nn.Flatten(),
    nn.Linear(in_features=12544, out_features=128),
    nn.ReLU(),
    nn.BatchNorm1d(num_features=128),# Dùng sau lớp Linear + activation/ không có
    nn.Linear(in_features=128, out_features=10)
)
```

```python
summary(model, input_size=(1,28,28),batch_size=32, device='cpu')
```
